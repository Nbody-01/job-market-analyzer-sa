# STEP 5 — Analyze Missing Values

## Purpose
Identify columns with missing data and quantify the extent.

## Method
Count and percentage of missing values per column.

```python
# Missing values analysis
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_report = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
})

print("Missing Values Report:")
print(missing_report[missing_report['Missing Count'] > 0].sort_values('Missing %', ascending=False))

if missing.sum() == 0:
    print("No missing values found.")
```

---

# STEP 6 — Analyze Duplicates

## Purpose
Identify duplicate rows and duplicate URLs.

## Method
Check for full-row duplicates and URL duplicates.

```python
# Duplicate analysis
full_duplicates = df.duplicated().sum()
id_duplicates = df.duplicated(subset=['Job_ID']).sum()
desc_duplicates = df.duplicated(subset=['Job_Description']).sum()

print("Duplicate Analysis:")
print(f"  Full row duplicates: {full_duplicates}")
print(f"  Job_ID duplicates: {id_duplicates}")
print(f"  Job_Description duplicates: {desc_duplicates}")

if full_duplicates > 0:
    print(f"\nDuplicate rows:")
    df[df.duplicated(keep=False)].sort_values(list(df.columns))
```

---

# STEP 7 — Inspect Unique Values

## Purpose
Understand the cardinality of categorical columns.

## Method
Count unique values for each categorical column.

```python
# Unique values for categorical columns
categorical_cols = ['Job_Title', 'Company', 'Industry', 'Province', 'City',
                     'Employment_Type', 'Education', 'Degree_Field', 'Remote',
                     'Contract_Type', 'Company_Size']

print("Unique Values per Categorical Column:")
print("-" * 50)
for col in categorical_cols:
    if col in df.columns:
        n_unique = df[col].nunique()
        print(f"  {col:20s}: {n_unique:4d} unique values")
```

---

# STEP 8 — Analyze Text Fields

## Purpose
Understand the text content (Job_Description and Skills).

## Method
Inspect sample text values and calculate text length statistics.

```python
# Text field analysis
print("Text Field Analysis:")
print("-" * 50)

# Job Description length
df['desc_length'] = df['Job_Description'].str.len()
print(f"\nJob_Description length:")
print(f"  Mean: {df['desc_length'].mean():.1f} characters")
print(f"  Median: {df['desc_length'].median():.1f} characters")
print(f"  Min: {df['desc_length'].min()} characters")
print(f"  Max: {df['desc_length'].max()} characters")

# Skills count
df['skills_count'] = df['Skills'].apply(lambda x: len(str(x).split(',')) if pd.notna(x) else 0)
print(f"\nSkills count per job:")
print(f"  Mean: {df['skills_count'].mean():.1f} skills")
print(f"  Median: {df['skills_count'].median():.1f} skills")
print(f"  Min: {df['skills_count'].min()} skills")
print(f"  Max: {df['skills_count'].max()} skills")

# Sample descriptions
print(f"\nSample Job Descriptions:")
for i, desc in enumerate(df['Job_Description'].head(3), 1):
    print(f"  {i}. {desc[:100]}...")
```

# STEP 12 — Define Cleaning Requirements

## Purpose
Based on the audit findings, define what needs to be cleaned.

## Cleaning Requirements

Based on the analysis above, the following cleaning steps are required:

1. **Remove duplicates** — Based on Job_ID and Job_Description
2. **Handle missing values** — Decide on imputation or removal per column
