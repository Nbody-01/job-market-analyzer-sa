# STEP 3 — Inspect Existing Collected Data

## Purpose
Understand what fields were collected and their completeness.

## Method
Check which expected fields are present and their fill rates.

```python
# Expected fields for job posting data
expected_fields = [
    'Job_ID', 'Job_Title', 'Company', 'Industry', 'Province', 'City',
    'Employment_Type', 'Experience_Years', 'Education', 'Degree_Field',
    'Remote', 'Contract_Type', 'Job_Description', 'Company_Size',
    'Skills'
]

print("Field Completeness Report:")
print("="*60)
print(f"{'Field':<20} {'Present':<10} {'Fill Rate':<12} {'Missing'}")
print("-"*60)

for field in expected_fields:
    if field in df.columns:
        present = "Yes"
        fill_rate = f"{(1 - df[field].isnull().sum() / len(df)) * 100:.1f}%"
        missing = df[field].isnull().sum()
    else:
        present = "No"
        fill_rate = "0.0%"
        missing = len(df)
    print(f"{field:<20} {present:<10} {fill_rate:<12} {missing}")
```

---

# STEP 4 — Validate Required Fields

## Purpose
Ensure all required fields for the project are present and usable.

## Method
Check if required fields exist and have sufficient data.

```python
# Required fields for the project
required_fields = {
    'Job_Title': 'Job title for role matching',
    'Company': 'Company name for recommendations',
    'Province': 'Location for geographic analysis',
    'Job_Description': 'Text for NLP and similarity analysis',
    'Skills': 'Skills for matching and gap analysis',
    'Education': 'Education requirements for matching',
    'Experience_Years': 'Experience requirements for matching',
    'Employment_Type': 'Job type for filtering',
    'Contract_Type': 'Contract type for filtering',
    'Remote': 'Work type for preference matching'
}

print("Required Fields Validation:")
print("="*60)
all_valid = True
for field, purpose in required_fields.items():
    if field in df.columns:
        non_null = df[field].notna().sum()
        status = "PASS" if non_null > 0 else "FAIL"
        if status == "FAIL":
            all_valid = False
        print(f"  [{status}] {field:<20} - {purpose} ({non_null}/{len(df)} non-null)")
    else:
        all_valid = False
        print(f"  [FAIL] {field:<20} - {purpose} (MISSING)")

print(f"\nOverall: {'ALL REQUIRED FIELDS PRESENT' if all_valid else 'SOME FIELDS MISSING'}")
```

# STEP 5 — Check Source Coverage

## Purpose
Understand the geographic and industry coverage of the dataset.

## Method
Analyze the distribution of provinces, industries, and companies.

```python
# Source coverage analysis
print("Source Coverage Analysis:")
print("="*60)

# Province coverage
print("\nProvince Coverage:")
province_counts = df['Province'].value_counts()
for prov, count in province_counts.items():
    pct = count / len(df) * 100
    print(f"  {prov:<20}: {count:>4} jobs ({pct:>5.1f}%)")

# Industry coverage
print("\nIndustry Coverage:")
industry_counts = df['Industry'].value_counts()
for ind, count in industry_counts.head(10).items():
    pct = count / len(df) * 100
    print(f"  {ind:<35}: {count:>4} jobs ({pct:>5.1f}%)")

# Company coverage
print(f"\nCompany Coverage:")
print(f"  Unique companies: {df['Company'].nunique()}")
print(f"  Top company: {df['Company'].value_counts().index[0]} ({df['Company'].value_counts().iloc[0]} jobs)")
```

# STEP 6 — Check Collection Quality

## Purpose
Assess the overall quality of the collected data.

## Method
Check for common collection issues like duplicates, incomplete records, and formatting problems.

```python
# Collection quality checks
print("Collection Quality Checks:")
print("="*60)

# Check for duplicates
full_dups = df.duplicated().sum()
id_dups = df.duplicated(subset=['Job_ID']).sum()
desc_dups = df.duplicated(subset=['Job_Description']).sum()

print(f"\nDuplicate Analysis:")
print(f"  Full row duplicates: {full_dups}")
print(f"  Job_ID duplicates: {id_dups}")
print(f"  Description duplicates: {desc_dups}")

# Check for empty strings
print(f"\nEmpty String Analysis:")
for col in ['Job_Title', 'Company', 'Job_Description', 'Skills']:
    if col in df.columns:
        empty_count = (df[col].astype(str).str.strip() == '').sum()
        print(f"  {col}: {empty_count} empty strings")

# Check for placeholder text
print(f"\nPlaceholder Text Analysis:")
placeholders = ['N/A', 'n/a', 'NA', 'null', 'NULL', 'None', '-']
for col in ['Job_Title', 'Company', 'Job_Description', 'Skills']:
    if col in df.columns:
        placeholder_count = df[col].astype(str).isin(placeholders).sum()
        print(f"  {col}: {placeholder_count} placeholder values")
```

# STEP 8 — Save Raw Dataset

## Purpose
Preserve the raw dataset for reproducibility.

## Method
Save a copy of the raw dataset with documentation.

```python
# Save raw dataset
df.to_csv('data/raw/sa_it_jobs_raw.csv', index=False)
print("Raw dataset saved to: data/raw/sa_it_jobs_raw.csv")

# Create data collection metadata
metadata = {
    'collection_date': '2024',
    'source': 'South African online job boards and recruitment sites',
    'total_records': len(df),
    'fields_collected': list(df.columns),
    'provinces_covered': df['Province'].nunique(),
    'companies_covered': df['Company'].nunique(),
    'industries_covered': df['Industry'].nunique(),
    'known_issues': {
        'full_duplicates': int(df.duplicated().sum()),
        'missing_values': int(df.isnull().sum().sum()),
        'description_duplicates': int(df.duplicated(subset=['Job_Description']).sum())
    },
    'ethical_considerations': [
        'Data collected from publicly available job postings',
        'No personal data of candidates included',
        'Respect for website terms of service and robots.txt',
        'Rate limiting applied during collection'
    ],
    'limitations': [
        'Snapshot of job market at collection time',
        'May not represent all South African IT jobs',
        'Some fields may have inconsistent formatting',
        'No salary information in dataset'
    ]
}

import json
with open('data/raw/collection_metadata.json', 'w') as f:
    json.dump(metadata, f, indent=2)

print("Collection metadata saved to: data/raw/collection_metadata.json")
```
