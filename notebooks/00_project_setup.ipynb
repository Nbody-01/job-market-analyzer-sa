# STEP 6 — Load Utility Functions

## Purpose
Define reusable utility functions that will be used across multiple notebooks.

## Method
Create a utility module with common functions.

```python
# Utility Functions
def save_dataframe(df, filepath, index=False):
    """Save a DataFrame to CSV."""
    df.to_csv(filepath, index=index)
    print(f"Saved: {filepath}")

def load_dataframe(filepath):
    """Load a DataFrame from CSV."""
    df = pd.read_csv(filepath)
    print(f"Loaded: {filepath} | Shape: {df.shape}")
    return df

def display_section_header(title):
    """Display a formatted section header."""
    print("\n" + "="*60)
    print(f"  {title}")
    print("="*60 + "\n")

def calculate_cramers_v(confusion_matrix):
    """Calculate Cramér's V statistic for categorical association."""
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    rcorr = r - ((r-1)**2)/(n-1)
    kcorr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

print("Utility functions loaded.")
```
